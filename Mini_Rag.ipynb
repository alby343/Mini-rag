```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

document = """
DifferentByte makes legacy systems intelligent. Enterprises connect old infrastructure to modern digital tools without writing code.
This saves engineering teams months of effort and reduces integration times by up to 90 percent.
The AI Developer role is based in Kochi and is full time.
The AI Developer builds backend scripts to integrate LLMs, vector databases and automated workflows into the flagship product.
Candidates should know Python, NumPy, Pandas and Scikit-Learn.
Familiarity with vector databases or RAG pipelines is a plus.
The selection process has an HR screening, a machine test and a final leadership interview.
"""

# Step 1: chunk by line
chunks = [line.strip() for line in document.split("\n") if line.strip()]
print(len(chunks), "chunks")

# Step 2: vectorize and retrieve
vectorizer = TfidfVectorizer(stop_words="english")
chunk_vectors = vectorizer.fit_transform(chunks)

def retrieve(question, k=2):
    q_vec = vectorizer.transform([question])
    scores = cosine_similarity(q_vec, chunk_vectors)[0]
    top = scores.argsort()[::-1][:k]
    return [(chunks[i], round(float(scores[i]), 3)) for i in top]

print(retrieve("Where is the AI Developer role based?"))
print(retrieve("What should candidates know?"))
```
